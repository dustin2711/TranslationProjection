```python
import cv2
import numpy as np
from file_watcher import FileWatcher
from ximea_image_capture import XimeaImageCapture
from config import Config
import matlike_utils
from enums import * 

subtractor_by_mode: dict[BackgroundSubtractorMode, cv2.BackgroundSubtractor] = {
    BackgroundSubtractorMode.MOG2: cv2.createBackgroundSubtractorMOG2(),
    BackgroundSubtractorMode.KNN: cv2.createBackgroundSubtractorKNN()
}

# fg_mask = cv2.normalize(fg_mask, None, 0, 255, cv2.NORM_MINMAX)
# _, fg_mask = cv2.threshold(fg_mask, 200, 255, cv2.THRESH_BINARY)
# cv2.waitKey(0)

CONFIG_PATH = r"C:\Users\sens\Desktop\FingernailProjection\FingernailProjection\config.json"
config = Config()
config.load_from_file(CONFIG_PATH)
config_watcher = FileWatcher(CONFIG_PATH)

image_capture = XimeaImageCapture()
while True:
    if config_watcher.has_timestamp_changed():
        config.load_from_file(CONFIG_PATH)

    image = image_capture(config)

   # Dynamically adjust subtractor and its parameters
    subtractor = subtractor_by_mode[config.subtraction_mode]
    subtractor.setHistory(config.history_count) 
    subtractor.setDetectShadows(config.detect_shadows)  
    match config.subtraction_mode:
        case BackgroundSubtractorMode.KNN:
            subtractor.setDist2Threshold(config.sensitivity_knn) 
        case BackgroundSubtractorMode.MOG2:
            subtractor.setVarThreshold(config.sensitivity_mog2)

    fg_mask = subtractor.apply(image)
    cv2.imshow("fg_mask", fg_mask)

    keyboard = cv2.waitKey(5)
    if keyboard == "q" or keyboard == 27:
        break

cv2.destroyAllWindows()
```
