```python
import sys
import os

sys.path.append(os.path.abspath(".."))
from stopwatch import stopwatch
import pygame
import cv2
import numpy as np

pygame.init()
screen = pygame.display.set_mode((3840, 2160), pygame.FULLSCREEN, display=1)


def cv_to_pygame_surface(cv_image):
    # Convert BGR (OpenCV) to RGB (Pygame)
    rgb_image = cv2.cvtColor(cv_image, cv2.COLOR_BGR2RGB)
    return pygame.surfarray.make_surface(np.rot90(rgb_image))


output_projection_image = cv2.imread(
    r"C:\Users\sens\Desktop\FingernailProjection\FingernailProjection\TestProjection.jpg"
)

while True:
    stopwatch.restart()

    pygame_surface = cv_to_pygame_surface(output_projection_image)
    screen.blit(pygame_surface, (0, 0))
    stopwatch.logandrestart("Convert")

    pygame.display.flip()
    stopwatch.logandrestart("Flip")

    for event in pygame.event.get():
        if event.type == pygame.QUIT or event.type == pygame.KEYDOWN:
            pygame.quit()
            exit()

    stopwatch.logandrestart("Event")
```
