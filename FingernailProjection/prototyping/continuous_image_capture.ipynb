```python
import sys
import os

sys.path.append(os.path.abspath(".."))
from cv_loop import CvLoop
from automatic_config_reloader import AutomaticConfigReloader
from file_watcher import FileWatcher
from config import Config
from image_display import ImageDisplay
from enums import ScaleMode
from enums import *
import cv2
from pylon_image_capture import PylonImageCapture

config, reloader = AutomaticConfigReloader.get_config_and_reloader(
    r"C:\Users\sens\Desktop\FingernailProjection\FingernailProjection\config.json", Config
)

capture = PylonImageCapture(config)

recorded_display = ImageDisplay(
    "Recorded Image",
    Anchor.TOP_LEFT,
    screen_index=1,  # This is the left screen with projector enabled
    scale_mode=ScaleMode.FILL_SCREEN_HEIGHT,
)


def step():
    reloader.reload_if_config_changed()
    image = capture.get_image()
    recorded_display.show(image)
    return image


CvLoop(step).start()

# Cleanup
cv2.destroyAllWindows()
del capture
```
