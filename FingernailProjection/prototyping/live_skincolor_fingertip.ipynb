```python
import sys
import os

sys.path.append(os.path.abspath(".."))
from skincolor_finger_apex_retriever import get_fingertip_data
from skin_detect_config import SkinDetectConfig
from image_display import ImageDisplay
from automatic_config_reloader import AutomaticConfigReloader
from config import Config
from drawing_utils import *
import matlike_utils
from enums import ScaleMode
import helper
from automatic_config_reloader import AutomaticConfigReloader
from config import Config
from image_display import ImageDisplay
from enums import ScaleMode
from enums import *
from cv_loop import CvLoop

config, config_reloader = AutomaticConfigReloader.get_config_and_reloader(
    r"C:\Users\sens\Desktop\FingernailProjection\FingernailProjection\config.json", Config
)

skin_config, skin_config_reloader = AutomaticConfigReloader.get_config_and_reloader(
    r"C:\Users\sens\Desktop\FingernailProjection\FingernailProjection\skin_config.json",
    SkinDetectConfig,
)

display = ImageDisplay(
    "Images", Anchor.TOP_LEFT, screen_index=1, scale_mode=ScaleMode.FILL_SCREEN_WIDTH
)

paths = helper.get_files_in_folder(
    r"C:\Users\sens\Desktop\FingernailProjection\FingernailProjection\prototyping\hand_dataset"
)

# Load images
images = [cv2.imread(path) for path in paths]
images = [images[0], images[1], images[3], images[8]]

shape = images[0].shape
fingertip_mark = (shape[1] // 2, shape[0] // 2)


def step() -> MatLike:
    config_reloader.reload_if_config_changed()
    skin_config_reloader.reload_if_config_changed()

    image_tuples = []

    for image in images:
        data = get_fingertip_data(image, skin_config, config, fingertip_mark)

        drawed_image = image.copy()
        draw_linestring(drawed_image, data.linestring, GREEN)
        draw_point(drawed_image, fingertip_mark, GREEN)
        # Draw circle
        draw_circle(drawed_image, data.center, data.radius)
        draw_point(
            drawed_image,
            (data.center[0], data.center[1] - data.radius),
            RED,
            3,
        )

        image_tuples.append(
            matlike_utils.merge_images_vertically(
                [
                    drawed_image,
                    data.mask,
                    data.mask_opened_closed,
                ]
            )
        )

    merged_image = matlike_utils.merge_images_horizontally(image_tuples)
    display.show(merged_image)

    return merged_image


CvLoop(step).start()

del display
```
